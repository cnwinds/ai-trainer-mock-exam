房屋交易与市场环境智能业务挖掘（填空已完成，可直接运行）


```python
import pandas as pd
import matplotlib.pyplot as plt

# 字体：考场有 SimHei；本环境可能无中文字体，失败则回退
try:
    plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
except Exception:
    pass
plt.rcParams['axes.unicode_minus'] = False

# 读取数据（与 notebook 同目录）
housing_data = pd.read_csv('housing_transactions.csv')
market_data = pd.read_csv('market_data.csv')

# 时间格式转换
housing_data['transaction_date'] = pd.to_datetime(housing_data['transaction_date'])
market_data['date'] = pd.to_datetime(market_data['date'])

# 业务机会点一：房屋定价优化
# 统计不同区域的平均房价，识别价值洼地和高溢价区域
avg_price_by_region = housing_data.groupby('region')['unit_price'].mean().sort_values(ascending=False)

print("房屋定价优化：不同区域的平均房价")
print(avg_price_by_region)
avg_price_by_region.plot(kind='bar', title='Avg unit price by region')
plt.xlabel('region')
plt.ylabel('avg unit_price')
plt.tight_layout()
plt.show()

# 业务机会点二：购房需求洞察
transaction_count_by_type = housing_data['house_type'].value_counts()

print("\n购房需求洞察：不同房型的交易数量")
print(transaction_count_by_type)
transaction_count_by_type.plot(kind='bar', title='Transaction count by house_type')
plt.xlabel('house_type')
plt.ylabel('count')
plt.tight_layout()
plt.show()

# 业务机会点三：市场风险预警
monthly_price = (
    housing_data.assign(month=housing_data['transaction_date'].dt.to_period('M').astype(str))
    .groupby('month', as_index=False)['unit_price']
    .mean()
    .rename(columns={'unit_price': 'avg_price'})
)

monthly_rate = (
    market_data.assign(month=market_data['date'].dt.to_period('M').astype(str))
    .groupby('month', as_index=False)['interest_rate']
    .mean()
)

risk_trend = pd.merge(monthly_price, monthly_rate, on='month', how='inner')

print("\n市场风险预警：房价与利率趋势")
print(risk_trend)
risk_trend.plot(x='month', y=['avg_price', 'interest_rate'], kind='line', marker='o',
                title='Price vs Interest Rate')
plt.xlabel('month')
plt.ylabel('value')
plt.tight_layout()
plt.show()
```
